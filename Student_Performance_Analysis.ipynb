```python

import pandas as pd

# Create our own student dataset
data = {
    "Student_ID": list(range(1, 21)),
    "Study_Hours": [2, 3, 4, 5, 2, 6, 7, 3, 5, 8,
                    4, 6, 7, 2, 9, 5, 3, 8, 6, 4],
    "Maths": [45, 55, 60, 70, 50, 80, 85, 55, 72, 90,
              65, 78, 88, 48, 95, 75, 58, 92, 82, 68],
    "Physics": [50, 58, 62, 72, 52, 78, 82, 60, 75, 88,
                67, 80, 85, 45, 93, 73, 55, 90, 84, 65],
    "Chemistry": [48, 60, 65, 68, 55, 82, 80, 58, 70, 92,
                  63, 76, 87, 50, 96, 78, 60, 94, 86, 70],
    "Attendance": [80, 85, 90, 92, 75, 95, 96, 82, 88, 98,
                   85, 93, 97, 78, 99, 90, 80, 96, 94, 87]
}

# Convert data into a DataFrame
df = pd.DataFrame(data)

# Display dataset
print(df)
```

```python

# Save dataset as CSV file
df.to_csv("student_performance.csv", index=False)

print("Dataset created and saved successfully!")
```

```python

# Calculate total marks
df["Total_Marks"] = df["Maths"] + df["Physics"] + df["Chemistry"]

# Calculate average marks
df["Average_Marks"] = df["Total_Marks"] / 3

# Display student performance
print(df[["Student_ID", "Total_Marks", "Average_Marks"]])
```

```python

# Statistical calculations

print("Mean Marks:", df["Average_Marks"].mean())

print("Median Marks:", df["Average_Marks"].median())

print("Mode Marks:", df["Average_Marks"].mode().tolist())

print("Highest Marks:", df["Average_Marks"].max())

print("Lowest Marks:", df["Average_Marks"].min())

print("Average Study Hours:", df["Study_Hours"].mean())
```

```python

# Calculate subject-wise averages

subjects = ["Maths", "Physics", "Chemistry"]

for subject in subjects:
    print(subject, "Average:", df[subject].mean())
```

```python

# Create performance categories

def performance(avg):
    if avg >= 85:
        return "Excellent"
    elif avg >= 70:
        return "Good"
    elif avg >= 50:
        return "Average"
    else:
        return "Needs Improvement"

df["Performance"] = df["Average_Marks"].apply(performance)

print(df[["Student_ID", "Average_Marks", "Performance"]])
```

```python

# Count performance categories

print(df["Performance"].value_counts())
```

```python

df.to_csv("student_performance.csv", index=False)

print("Updated dataset saved successfully!")
```

```python

import matplotlib.pyplot as plt
import seaborn as sns

# Set chart style
sns.set_style("whitegrid")

print("Visualization libraries are ready!")
```

```python

# Calculate subject averages
subject_avg = df[["Maths", "Physics", "Chemistry"]].mean()

# Create bar chart
plt.figure(figsize=(8, 5))

plt.bar(subject_avg.index, subject_avg.values,
        color=["skyblue", "orange", "lightgreen"])

plt.title("Subject-wise Average Marks")
plt.xlabel("Subjects")
plt.ylabel("Average Marks")

plt.show()
```

```python

# Sort students by study hours
line_data = df.sort_values("Study_Hours")

plt.figure(figsize=(9, 5))

plt.plot(line_data["Study_Hours"],
         line_data["Average_Marks"],
         marker="o",
         color="blue")

plt.title("Study Hours vs Average Marks")
plt.xlabel("Study Hours")
plt.ylabel("Average Marks")

plt.grid(True)
plt.show()
```

```python

# Count performance categories
performance_counts = df["Performance"].value_counts()

plt.figure(figsize=(7, 7))

plt.pie(performance_counts.values,
        labels=performance_counts.index,
        autopct="%1.1f%%",
        startangle=90)

plt.title("Student Performance Distribution")

plt.show()
```

```python

plt.figure(figsize=(8, 5))

plt.hist(df["Average_Marks"],
         bins=5,
         color="skyblue",
         edgecolor="black")

plt.title("Distribution of Student Average Marks")
plt.xlabel("Average Marks")
plt.ylabel("Number of Students")

plt.show()
```
